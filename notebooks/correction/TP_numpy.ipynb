<div style="padding: 12px;"> Créer une fonction « gazparfait » permattant de modéliser PV = nRT avec les unités suivantes (généralement utilisées):<br> - P = pression du gaz en Pa<br> - V = volume en m3<br> - n = quantité de matière en mol<br> - T = température absolue en K (Kelvins)<br> - R = 8.314 SI (si les unités utilisées sont celles ci-dessus)<br><br> Calculer l'array des valeurs de quantité de matière (n) Y associées aux volumes X = [15, 89, 56, 78, 152, 66, 48, 77, 2, 96]. On se placera dans les conditions suivantes :<br> - P = 101 325 Pa ( = 1 atm) <br> - T = 20°C <br> </div>

```python
def gazparfait(V, P, R, T):
    n = (P*V)/(R*T)
    return(n)

P = 101325 # 1 atm
T = 293.15 # il fait 20 degrés
R = 8.314
X = np.array([15, 89, 56, 78, 152, 66, 48, 77, 2, 96])

gazparfait(X, P, R, T)
```

```python
rng = np.random.default_rng()
u_01 = rng.uniform(0,1,1000)
n_02 = rng.normal(0,np.sqrt(2),1000)

print("Loi U(0,1)")
print("Moyenne :", np.mean(u_01))
print("Variance :", np.var(u_01))

print("\nLoi N(0,2)")
print("Moyenne :", np.mean(n_02))
print("Variance :", np.var(n_02))
```

```python
def statdesc(x):
    return {
        "moyenne": np.mean(x),
        "mediane": np.median(x),
        "ecart_type": np.std(x),
        "minimum": np.min(x),
        "maximum": np.max(x)
    }
statdesc(u_01)
```

```python
import matplotlib.pyplot as plt

# Q1
X = rng.uniform(size = (10, 2))
```

```python
#Q2
print(X[:,0])
print(X[:,1])
fig = plt.figure()
plt.scatter(X[:, 0], X[:, 1], s=100)
plt.show()
```

```python
#Q3
print(X.shape)
X1 = X[:, np.newaxis, :]
print("dimension X1 : ", X1.shape)
X2 = X[np.newaxis, :,:]
print("dimension X2 : ", X2.shape)
dist1 = (X1 - X2) ** 2
print("dimension dist1 : ", dist1.shape)
dist = np.sqrt(np.sum((X1 - X2) ** 2, axis=-1))
print("dimension dist : ", dist.shape)
```

```python
print(dist)
print(dist.diagonal())
```

```python
#Q4
np.argsort(dist)
```

```python
K = 2
nearest_partition = np.argpartition(dist, K + 1, axis=1)
print(nearest_partition)
```

```python
fig = plt.figure()
for i in range(X.shape[0]):
    for j in nearest_partition[i, :K+1]:
        # plot a line from X[i] to X[j]
        # use some zip magic to make it happen:
        plt.plot(*zip(X[j], X[i]), color='black')
fig
plt.savefig("knn.png", bbox_inches='tight')
```
