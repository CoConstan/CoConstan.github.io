```python
import folium
import pandas as pd
from folium.plugins import MarkerCluster

velov_data = "https://www.data.gouv.fr/api/1/datasets/r/839491c5-35a0-4520-b4e5-1552fe5d9d68"
stations = pd.read_csv(velov_data)
```

```python
# Coordonnées de Lyon
lyon_coords = [45.7640, 4.8357]

# Création de la carte
carte_lyon = folium.Map(
    location=lyon_coords,
    zoom_start=13,
    tiles="OpenStreetMap"
)
```

```python
# Carte centrée sur Lyon
carte_lyon = folium.Map(
    location=[45.7640, 4.8357],
    zoom_start=13,
    tiles="OpenStreetMap"
)

# Cluster
marker_cluster = MarkerCluster().add_to(carte_lyon)

for _, row in stations.iterrows():
    popup_text = f"""
    <b>{row['name']}</b><br>
    Vélos disponibles : {row['available_bikes']}<br>
    Capacité totale : {row['bike_stands']}
    """

    folium.Marker(
        location=[row['lat'], row['lng']],
        popup=popup_text,
        tooltip=row['name']
    ).add_to(marker_cluster)

carte_lyon
```

```python
carte_lyon.save("carte_velov_lyon.html")
```
